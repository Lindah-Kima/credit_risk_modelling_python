#### Preprocessing continous variables

```python
#display values that a column takes
loan_data['emp_length'].unique() 
```

```python
#str.replace (<string to be replaced>, <new replacement string>)  - finds a string inside of another string and if found replace first string with given new string , if the first string is replaced with nothing it will be deleted from the given string
loan_data['emp_length_int'] = loan_data['emp_length'].str.replace(r'\+', ' ', regex=True)
loan_data['emp_length_int'] = loan_data['emp_length_int'].str.replace('< 1 year', str(0))
loan_data['emp_length_int'] = loan_data['emp_length_int'].str.replace('n/a',  str(0))
loan_data['emp_length_int'] = loan_data['emp_length_int'].str.replace(' years', '')
loan_data['emp_length_int'] = loan_data['emp_length_int'].str.replace(' year', '')
# We store the preprocessed ‘employment length’ variable in a new variable called ‘employment length int’,
# We assign the new ‘employment length int’ to be equal to the ‘employment length’ variable with the string ‘+ years’
# replaced with nothing. Next, we replace the whole string ‘less than 1 year’ with the string ‘0’.
# Then, we replace the ‘n/a’ string with the string ‘0’. Then, we replace the string ‘space years’ with nothing.
# Finally, we replace the string ‘space year’ with nothing.
```

```python
#transfomr it to numeric column
loan_data['emp_length_int'] = pd.to_numeric(loan_data['emp_length_int'])
```

#### Preprocessing Discrete variables

##### Creating dummy variables (one-hot encoding)

```python
loan_data.info()
```

```python
# Create dummy variables for categorical var - K-1 dummy var to represent the info about k categories
# pd.get_dummies(loan_data['grade'])
pd.get_dummies(loan_data['grade'], dtype=int)
```

```python
pd.get_dummies(loan_data['grade'], prefix = 'grade', prefix_sep = ' : ', dtype= int) 
```

```python
loan_data_dummies = [pd.get_dummies(loan_data['grade'], prefix = 'grade', prefix_sep = ' : ', dtype=int),
                    pd.get_dummies(loan_data['sub_grade'], prefix = 'sub_grade', prefix_sep = ' : ', dtype=int),
                    pd.get_dummies(loan_data['home_ownership'], prefix = 'home_ownership', prefix_sep = ' : ',dtype=int),
                    pd.get_dummies(loan_data['verification_status'], prefix = 'verification_status', prefix_sep = ' : ',dtype=int),
                    pd.get_dummies(loan_data['loan_status'], prefix = 'loan_status', prefix_sep = ' : ',dtype=int),
                    pd.get_dummies(loan_data['purpose'], prefix = 'purpose', prefix_sep = ' : ', dtype=int),
                    pd.get_dummies(loan_data['addr_state'], prefix = 'addr_state', prefix_sep = ' : ',dtype=int),
                    pd.get_dummies(loan_data['initial_list_status'], prefix = 'initial_list_status', prefix_sep = ' : ',dtype=int)]
```

```python
#make it a df
loan_data_dummies = pd.concat(loan_data_dummies, axis =1)
```

```python
type(loan_data_dummies)
```

```python
#combine dummies to original df by rows axis = 0 or columns axis = 1
loan_data = pd.concat([loan_data, loan_data_dummies], axis = 1)
```

##### Checking and cleaning missing values

```python
#returns true if theres a missing value else false
loan_data.isnull() 
```

```python
#sum the boolean values to get the sum of missing values by variable
pd.options.display.max_rows = None
loan_data.isnull().sum()
```

```python
#Imputing missing variables
pd.options.display.max_rows = 100
```

```python
# loan_data['total_rev_hi_lim'].fillna(loan_data ['funded_amnt'], inplace = True )
loan_data['total_rev_hi_lim'] = loan_data['total_rev_hi_lim'].fillna(loan_data ['funded_amnt'])
```

```python
loan_data['total_rev_hi_lim'].isnull().sum()
```

```python
# loan_data['annual_inc'].fillna(loan_data ['funded_amnt'], inplace = True)
```

```python
# We fill the missing values with the mean value of the non-missing values.
loan_data['annual_inc'] = loan_data['annual_inc'].fillna(loan_data['annual_inc'].mean())
```

```python
# We fill the missing values with zeroes.
loan_data['mths_since_earliest_cr_line'] = loan_data['mths_since_earliest_cr_line'].fillna(0)
```

```python
# List of columns that need to be filled with 0
cols_to_fix = [
    'emp_length_int', 'delinq_2yrs', 'inq_last_6mths', 
    'open_acc', 'pub_rec', 'total_acc', 
    'mths_since_earliest_cr_line', 'acc_now_delinq'
]

# Update the dataframe in one clean step
loan_data[cols_to_fix] = loan_data[cols_to_fix].fillna(0)
```
